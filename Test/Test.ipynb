```python
import os
import torch
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

device = torch.accelerator.current_accelerator().type if torch.accelerator.is_available() else "cpu"
print(f"Using {device} device")


class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(3, 5),
            nn.ReLU(),
            nn.Linear(5, 1),
            nn.ReLU(),
        )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)
        return logits


model = NeuralNetwork().to(device)
print(model)


X = torch.rand(1, 3, device=device)
logits = model(X)
pred_probab = nn.Softmax(dim=1)(logits)
y_pred = pred_probab.argmax(1)
print(f"Predicted class: {y_pred}")


input_image = torch.rand(1,3)
print(input_image.size())
```

# Question 2

```python
from transformers import pipeline

paragraph = (
    "Rohit Sharma is a indian mens cricket team captian. "
    "He won us T20 world cup and Campions Trophy."
    "He born in Nagpur on 31 of July."
    "His full name is Rohit Gururnath Sharma."
)

summarizer = pipeline("summarization")

summary_result = summarizer(paragraph, max_length=30, min_length=10, do_sample=False)

print("--- Original Text ---")
print(paragraph)

print("\n--- Generated Summary ---")
print(summary_result[0]['summary_text'])
```

# LLM

# Question 1



*   In first pass , the first token is generated (llms) then the next token is predicted by processing the first token.
*   while second pass (generated) token is predicted then on the basis of 'llms','generated' nest token is predicted (text) .


*   This loop continues until the whole sentence is generated.







*   Generating long sentences takes too time to process because they need to process all the past tokens to predict the next token.
*   Instead of processing the same past tokens we use different techniques like KV Cache , etc to reduce the timing of processing .

    

*   KV Cache save all the past tokens/information in a temporary storage and retrive it when required.
